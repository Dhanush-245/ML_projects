```python
# =====================================================
# Loan Approval Prediction Project
# Step 4 : Data Cleaning
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("/Users/lingareddydhanushkumar/Desktop/Codex/ML/01_Loan_Approval_Prediction/data/LoanApprovalPrediction.csv")

print("Dataset Loaded Successfully!")
```

```python
df.head()
```

```python
loan_df = df.copy()

loan_df.head()
```

```python
print("Rows :", loan_df.shape[0])
print("Columns :", loan_df.shape[1])
```

```python
loan_df.isnull().sum()
```

```python
missing = pd.DataFrame({
    "Missing Values": loan_df.isnull().sum(),
    "Percentage": round(loan_df.isnull().mean() * 100, 2)
})

missing[missing["Missing Values"] > 0]
```

```python
loan_df.drop("Loan_ID", axis=1, inplace=True)

loan_df.head()
```

```python
print("Duplicate Rows :", loan_df.duplicated().sum())
loan_df.drop_duplicates(inplace=True)
```

```python
num_cols = loan_df.select_dtypes(include=["int64", "float64"]).columns

print(num_cols)
```

```python
cat_cols = loan_df.select_dtypes(include="object").columns

print(cat_cols)
```

```python
loan_df["Dependents"].fillna(
    loan_df["Dependents"].mode()[0],
    inplace=True
)
```

```python
loan_df["LoanAmount"].fillna(
    loan_df["LoanAmount"].median(),
    inplace=True
)
```

```python
loan_df["Loan_Amount_Term"].fillna(
    loan_df["Loan_Amount_Term"].mode()[0],
    inplace=True
)
```

```python
loan_df["Credit_History"].fillna(
    loan_df["Credit_History"].mode()[0],
    inplace=True
)
```
