# Fraud / Anomaly Detection
## Notebook 02 — Data Cleaning

### Objective

Clean and validate the transaction dataset while preserving potentially meaningful anomalous observations.

### Important Principle

An unusual transaction is not automatically a bad data point.

Because this is an anomaly detection project, potential outliers must be preserved for later analysis.
