# Fraud / Anomaly Detection
## Notebook 11 — PCA Anomaly Visualization

### Objective
Project high-dimensional transactions into two dimensions and visualize fraud and anomaly scores.

## Imports and sample

```python
from pathlib import Path
import sys, numpy as np, pandas as pd
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.fraud_detection import load_features_labels, evaluate_scores, save_results
X, y = load_features_labels(ROOT)
RANDOM_STATE = 42
print('Features:', X.shape, '| fraud:', int(y.sum()))
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.decomposition import PCA
from src.fraud_detection import stratified_working_sample
Xw, yw = stratified_working_sample(X, y, max_rows=20000, random_state=RANDOM_STATE)
```

**Interpretation.** PCA is used for visualization, not as proof of separability.

**ML decision.** Preserve labels only for plot coloring.

## Project and plot

```python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
points = pca.fit_transform(Xw)
plot_df = pd.DataFrame({'PC1': points[:,0], 'PC2': points[:,1], 'Class': yw.to_numpy()})
plt.figure(figsize=(10,7)); sns.scatterplot(data=plot_df, x='PC1', y='PC2', hue='Class', alpha=.45, s=18)
plt.title(f'PCA projection — explained variance {pca.explained_variance_ratio_.sum():.1%}'); plt.show()
```

**Interpretation.** Overlap is expected: PCA preserves global variance, while fraud may be locally or nonlinearly anomalous.

**ML decision.** Do not select a model from this plot alone.
